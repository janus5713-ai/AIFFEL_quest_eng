```python
class Square:
    def __init__(self):
        self.square = int(input('넓이를 구하고 싶은 사각형의 숫자를 써주세요.\n 1.직사각형 2.평행사변형 3.사다리꼴 \n >>>'))

        if self.square == 1:
            print('직사각형 함수는 rect()입니다.')
            print(self.rect())

        elif self.square == 2:
            print('평행사변형 함수는 par()입니다.')
            print(self.par())
        
        elif self.square == 3:
            print('사다리꼴 함수는 trape()입니다.')
            print(self.trape())
        
        else:
            print('1, 2, 3 중에서 다시 입력해주세요')

    def rect(self):
        width, vertical = map(int, input('가로, 세로를 입력하세요. 예시 : 가로,세로\n >>>').split(','))
        area = width * vertical
        result = '직사각형의 넓이는 : ' + str(area)
        return result

    def par(self):
        base, height = map(int, input('밑변, 높이를 입력하세요.').split(','))
        area = base * height
        result = '평행사변형의 넓이는 : ' +str(area)
        return result

    def trape(self):
        top, bottom, height  = map(int, input('윗변, 밑변, 높이를 입력하세요').split(','))
        area = (top + bottom) * height / 2
        result = '사다리꼴의 넓이는 : ' +str(area)
        return result

stop = True #상태지정 변수

while stop:
    while True:
        Square()
        again = input("넓이를 더 계산하시겠습니까? (y/n): ")

        if again in ['y', 'Y', 'ㅛ']:
            break
            Square
    
        elif again in ['n', 'N', 'ㅜ']:
            stop = False
            break
        else :
            continue #질문 루프     
```

```python
class Kiosk:   #class 생성

    def __init__(self):

        menu = ['에스프레소', '아메리카노', '초코라떼', '카페모카', '라떼',
                 '그린티라떼', '유자차', '캐모마일' , '얼그레이' ]
        price = [2000, 2500, 3500, 3000, 2500, 3500, 3000, 3500, 3500]      
        #메뉴, 가격 메서드 생성
        self.menu = menu  
        self.price = price
        self.order = []

    # 메뉴 출력 메서드
    def menu_print(self):
        for i in range(len(self.menu)):
            print(i + 1, self.menu[i], ' : ', self.price[i], '원')

    # 주문 메서드
    def menu_select(self):
        print()  
        
        while True:
            n = 0
            
            while n < 1 or len(self.menu) < n:
                n = int(input("음료 번호를 입력하세요 : "))  # 음료 번호 입력

                # 메뉴판에 있는 음료 번호일때
                if 1 <= n and n <= len(self.menu):
                    self.price_sum = self.price[n-1]  # 선택 음료의 가격
            
                # 메뉴판에 없는 번호일 때
                else:  
                    print("없는 메뉴입니다. 다시 주문해 주세요.")

            # 음료 온도 물어보기
            t = 0  # 기본값을 넣어주고
            while t != 1 and t != 2:  # 1이나 2를 입력할 때까지 물어보기
                t= int(input("HOT 음료는 1을, ICE 음료는 2를 입력하세요 : "))
                #  음료의 온도에 따라 temp 변수를 "HOT" 또는 "ICE"로 지정하세요.   
                    
                if t == 1:
                    self.temp = 'HOT'
                    self.order.append([self.temp, self.menu[n-1], self.price[n-1]])
                    break
                elif t == 2:
                    self.temp = 'ICE'
                    self.order.append([self.temp, self.menu[n-1], self.price[n-1]])
                    break
                else:    
                    print("1과 2 중 하나를 입력하세요.\n")
            
            print('주문 음료', self.temp, self.menu[n-1], ' : ', self.price[n-1], '원')  # 온도 속성을 추가한 주문 결과 출력
            #주문 이후에 결제를 해야함
            
            추가주문 = input("추가 주문 하시겠습니까? (1.예 2.아니오) : ")
            if 추가주문 == '1' :
                self.menu_print()
            if 추가주문 == '2' :
                break # 추가주문 X
        총금액 = 0

        #self.order에 담긴 음료들의 가격을 모두 더함
        for drink in self.order:
            총금액 += int(drink[2]) 

        self.pay(총금액)
        
    # 지불 메서드
    def pay(self, dollar): #매개변수
        
        print(f"지불할 금액은 {dollar}원입니다.") #f-string
        #변수지정
        ca = input('지불 방식을 선택해주세요(1.현금, 2.카드): ')

        if ca in ['1', 'cash', '현금']: #현금 방식 리스트
            print('직원을 호출하겠습니다.')
        elif ca in ['2','card', '카드']: #카드 방식 리스트
            print('IC칩 방향에 맞게 카드를 꽂아주세요')
        else:
            print('다시 결제를 시도해 주세요.')

        self.order_list()

    # 주문서 출력 메서드 
    def order_list(self):
        print('\n')
        print('-' * 44)
        print("품명                      단가  수량    금액")
        print('-' * 44)
        
        총합계 = 0
        
        for drink in self.order:
            온도 = drink[0]
            이름 = drink[1]
            단가 = int(drink[2]) 
            수량 = 1
            금액 = 단가 * 수량
            총합계 += 금액
            
            품명 = f"({온도}){이름}"
            실제화면길이 = len(품명.encode('cp949')) #계산서 삐져나가서 열심히 검색함...
            
            
            남는공간 = 22 - 실제화면길이
            여백 = ' ' * 남는공간
            
            
            print(f"{품명}{여백}{단가:>8}{수량:>6}{금액:>8}")
            
        print('-' * 44)
        
        print(f"합계 금액 : {총합계:>28}원")
        print('-' * 44)
        
my_kiosk = Kiosk()
my_kiosk.menu_print()
my_kiosk.menu_select()
```
