```python
[퀘스트 1번: 은행계좌 만들기]

Q1. Account 클래스 : 은행에 가서 계좌를 개설하면 은행이름, 예금주, 계좌번호, 잔액이 설정됩니다. Account 클래스를 생성한 후 생성자(hint: 매직메서드...!!)를 구현해보세요. 생성자에서는 예금주와 초기 잔액만 입력 받습니다. 은행이름은 SC은행으로 계좌번호는 3자리-2자리-6자리 형태로 랜덤하게 생성됩니다. (은행이름: SC은행, 계좌번호: 111-11-111111)

Q2. 클래스 변수 : 클래스 변수를 사용해서 Account 클래스로부터 생성된 계좌 객체의 개수를 저장하세요.

Q3. 클래스 변수 출력 : Account 클래스로부터 생성된 계좌의 개수를 출력하는 get_account_num() 메서드를 추가하세요.

Q4. 입금 메서드 : Account 클래스에 입금을 위한 deposit 메서드를 추가하세요. 입금은 최소 1원 이상만 가능합니다.

Q5. 출금 메서드 : Account 클래스에 출금을 위한 withdraw 메서드를 추가하세요. 출금은 계좌의 잔고 이상으로 출금할 수는 없습니다.

Q6. 정보 출력 메서드 : Account 인스턴스에 저장된 정보를 출력하는 display_info() 메서드를 추가하세요. 잔고는 세자리마다 쉼표를 출력하세요.

(은행이름: SC은행, 예금주: 파이썬, 계좌번호: 111-11-111111, 잔고: 10,000원)

Q7. 이자 지급하기 : 입금 횟수가 5회가 될 때 잔고를 기준으로 1%의 이자가 잔고에 추가되도록 코드를 변경해보세요.

Q8. 여러 객체 생성 : Account 클래스로부터 3개 이상 인스턴스를 생성하고 생성된 인스턴스를 리스트에 저장해보세요.

Q9. 객체 순회 반복문을 통해 리스트에 있는 객체를 순회하면서 잔고가 100만원 이상인 고객의 정보만 출력하세요.

Q10. 입금과 출금 내역이 기록되도록 코드를 업데이트 하세요.

(입금 내역과 출금 내역을 출력하는 deposit_history와 withdraw_history 메서드를 추가하세요.)
```

```python
import random

class Account:
    account_score = 0

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        self.bank = "sc은행"

        Account.account_score += 1
        self.deposit_count = 0
        self.deposit_log = [] #입금기록
        self.withdraw_log = [] #출금기록
        
    #계좌번호 생성... 랜덤이니까 import random 
    #zfill 함수 > 문자열 왼쪽에 0을 채워줌 ex)1자리 숫자가 나왔을 경우 001,002.... 
    #하지만 문자열 전용 메서드 이기 때문에 숫자를 문자열로 변환해야함 str.zfill

        num1 = str(random.randint(0,999)).zfill(3)
        num2 = str(random.randint(0,99)).zfill(2)
        num3 = str(random.randint(0,999999)).zfill(6)
        self.account_num = num1+'-'+num2+'-'+num3 #문자열로 변환했고 계좌의 -를 넣으려면 연산자가 아닌 문자열로 넣어야함

    #Account 클래스로부터 생성된 계좌의 개수를 출력하는 get_account_num() 메서드
    def get_account_num(self):
        print(Account.account_score)

    def deposit(self,amount):
        if amount >=1: #잔액이 1원 이상 일 때
            self.balance += amount
            self.deposit_count += 1
            self.deposit_log.append(amount) #입금내역 추가 #리스트에 append로 넣기             
            
            if self.deposit_count % 5 == 0:
                self.balance += self.balance *0.01 #잔고 * 1% 이자율
        else:
            print("입금액은 1원 이상 이어야합니다.")

    def withdraw(self, amount):
        #잔액이 잔고보다 작거나 같다면 출금
        if amount <= self.balance:
            self.balance -= amount
            self.withdraw_log.append(amount)
        else :
            print("출금 가능한 잔액이 부족합니다.")
    
    def display_info(self): #정보 출력 은행이름,예금주,잔액,계좌번호
        print(f"은행이름 : {self.bank}")
        print(f"예금주 : {self.name}")
        print(f"잔액 : {self.balance:,}원") #:, <<숫자 3단위씩 끊는 기능이 있음 :, 반드시 붙여서 써야함
        print(f"계좌번호 : {self.account_num}")

List = []

#100만원 이상 보유자, 미만 보유자 생성
a = Account("김", 1000000)
b = Account("이", 2000000)
c = Account("박", 1500000)
d = Account("박", 150000)

#append로 추가
List.append(a)
List.append(b)
List.append(c)
List.append(d)

for account in List:
    if account.balance >= 1000000:  #잔고 100만원이상
       account.display_info() #출력
```
